## Айнымалылар

**Мақсаты:** айнымалы ұғымын түсіну, оған мән меншіктеу, мәнін өзгерту және дұрыс атау беру ережелерін меңгеру.

Айнымалы — компьютер жадында мәнді сақтайтын атаулы орын. Python-да айнымалы `=` таңбасы арқылы мән меншіктелген сәтте пайда болады, оның типін алдын ала көрсету қажет емес. Математикадан айырмашылығы, бағдарламалауда `=` теңдікті емес, оң жақтағы мәнді сол жақтағы айнымалыға жазуды білдіреді.

### Айнымалының мәнін өзгерту

Айнымалының мәні тұрақты емес: оған жаңа мән меншіктегенде ескі мән жойылады. Сондай-ақ айнымалының жаңа мәнін оның ескі мәні арқылы есептеуге болады.

```python
# Жаңа мән меншіктеу ескі мәнді ауыстырады
age = 20
print("Жаңа жас:", age)
```

### Бір уақытта бірнеше айнымалыға мән беру

Python бір жолда бірнеше айнымалыға мән меншіктеуге мүмкіндік береді. Бұл тәсілмен екі айнымалының мәнін оңай алмастыруға болады.

```python
# Бір жолда бірнеше айнымалыға әртүрлі мән беру
x, y, z = 1, 2, 3
print(x, y, z)
```

```python
# Екі айнымалының мәнін алмастыру
x, y = y, x
print("x =", x)
print("y =", y)
```

## Деректер типтері

Әр мәннің өз типі болады, ал тип сол мәнмен қандай амалдар жасауға болатынын анықтайды. Негізгі қарапайым типтер: `int` — бүтін сан, `float` — бөлшек сан, `str` — мәтін (жол), `bool` — логикалық мән. Мәннің типін `type()` функциясы көрсетеді.

## Сандармен жұмыс

**Мақсаты:** бүтін және бөлшек сандармен арифметикалық амалдарды орындауды, амалдардың орындалу ретін түсінуді үйрену.

Python-да сандардың екі негізгі типі бар: `int` (бүтін сандар) және `float` (ондық бөлшегі бар сандар). Сандармен қосу, азайту, көбейту, бөлу, бүтін бөлу, қалдық табу және дәрежеге шығару амалдары орындалады. Амалдардың орындалу реті математикадағыдай: алдымен жақша, содан кейін дәреже, көбейту мен бөлу, ең соңында қосу мен азайту.

### Арифметикалық амалдар

```python
a = 17
b = 5
```

```python
# Қосу, азайту, көбейту
print(a + b)
print(a - b)
print(a * b)
```

```python
# Кәдімгі бөлу: нәтиже әрқашан float типті
print(a / b)
print(10 / 2)
```

```python
# Бүтін бөлу және бөлудің қалдығы
print(a // b)   # бөліндінің бүтін бөлігі
print(a % b)    # бөлудің қалдығы
```

```python
# Дәрежеге шығару
print(2 ** 10)
print(9 ** 0.5)   # 0.5 дәрежесі квадрат түбірге тең
```

Қалдық табу амалы `%` іс жүзінде жиі қолданылады. Мысалы, санның жұп немесе тақ екенін тексеру үшін оның 2-ге бөлгендегі қалдығы қарастырылады.

```python
# Жұп санның 2-ге бөлгендегі қалдығы 0, тақ санның қалдығы 1
print(8 % 2)
print(7 % 2)
```

### Амалдардың орындалу реті

```python
# Көбейту қосудан бұрын орындалады
print(2 + 3 * 4)
```

```python
# Жақша амалдардың ретін өзгертеді
print((2 + 3) * 4)
```

### Құрама меншіктеу операторлары

Айнымалының мәнін өзгертіп, нәтижені сол айнымалыға қайта жазу үшін қысқартылған операторлар қолданылады: `+=`, `-=`, `*=`, `/=`.

```python
balance = 1000
balance += 500   # balance = balance + 500
print(balance)
```

```python
balance -= 200   # balance = balance - 200
print(balance)

balance *= 2     # balance = balance * 2
print(balance)
```

### Сандарға арналған пайдалы функциялар

```python
# Абсолют шама және дөңгелектеу
print(abs(-15))
print(round(3.14159, 2))   # үтірден кейін 2 таңбаға дейін дөңгелектеу
print(round(7.6))          # бүтін санға дейін дөңгелектеу
```

```python
# Ең үлкен және ең кіші мәнді табу
print(max(4, 18, 7))
print(min(4, 18, 7))
```

### Шағын мысал: орташа баллды есептеу

Үйренген амалдарды қолданып, үш пәннен алынған баллдардың орташа мәнін есептейік.

```python
# Үш пәннің баллдары
math_score = 85
physics_score = 78
python_score = 92
```

```python
# Орташа мәнді есептеу және дөңгелектеу
average = (math_score + physics_score + python_score) / 3
print("Орташа балл:", round(average, 1))
```

## Жолдар (мәтіндік деректер)

**Мақсаты:** мәтіндік деректермен жұмыс істеуді: жолдарды біріктіруді, символдарға қол жеткізуді, жолдың бөлігін алуды және жол әдістерін қолдануды үйрену.

Жол (`str`) — тырнақшаға алынған символдар тізбегі. Жолдың әр символының өз реттік нөмірі, яғни индексі бар, ал индекс 0-ден басталады. Жолдар өзгермейтін тип болып табылады: жол әдістері бастапқы жолды өзгертпейді, жаңа жол қайтарады.

### Жолдарды біріктіру және қайталау

```python
first_name = "Aigerim"
last_name = "Sadykova"
```

```python
# + операторы жолдарды біріктіреді
full_name = first_name + " " + last_name
print(full_name)
```

```python
# * операторы жолды қайталайды
print(last_name * 30)
print(last_name * 3)
```

### Индекстеу

Жолдың жеке символына квадрат жақша ішінде индексін көрсету арқылы қол жеткізіледі. Теріс индекс жолдың соңынан бастап санайды: `-1` соңғы символды білдіреді.

```python
word = "Python"
```

```python
# Оң индекстер: 0-ден басталады
print(word[0])
print(word[1])
```

```python
# Теріс индекстер: соңынан санау
print(word[-1])
print(word[-2])
```

```python
# len() функциясы жолдың ұзындығын қайтарады
print(len(word))
```

### Кесінді алу (slicing)

`[start:stop]` жазбасы жолдың `start` индексінен басталып, `stop` индексіне дейінгі (`stop` кірмейді) бөлігін қайтарады. Шекараның бірін көрсетпесе, жолдың басы немесе соңы алынады.

```python
# 0, 1, 2 индекстегі символдар
print(word[0:3])
```

```python
# Басынан бастап және соңына дейін
print(word[:2])
print(word[2:])
```

```python
# Қадаммен кесу: [::-1] жолды кері айналдырады
print(word[::2])
print(word[::-1])
```

### Жол әдістері

Әдіс — белгілі бір типке тиесілі функция, ол нүкте арқылы шақырылады. Жолдарда регистрді өзгерту, бос орындарды тазалау, мәтінді ауыстыру және бөлу сияқты көптеген әдістер бар.

```python
text = "  Python Programming  "
```

```python
# Регистрді өзгерту
print(text.upper())
print(text.lower())
```

```python
# Басы мен соңындағы бос орындарды алып тастау
clean = text.strip()
print(clean)
```

```python
# Мәтін бөлігін ауыстыру
print(clean.replace("Programming", "Basics"))
```

```python
# Жолды бөліктерге бөлу: нәтиже тізім болады
sentence = "apple,banana,cherry"
print(sentence.split(","))
```

```python
# Символдар санын және орнын табу
print(clean.count("o"))
print(clean.find("Prog"))
```

### Жолда ішкі жолдың бар-жоғын тексеру

`in` операторы бір жолдың екінші жолдың ішінде бар-жоғын тексереді және логикалық мән (`True` немесе `False`) қайтарады.

```python
email = "student@university.kz"
print("@" in email)
print(".com" in email)
```

### Форматталған жолдар (f-жолдар)

Тырнақшаның алдына `f` әрпі қойылса, фигуралық жақша `{}` ішіне айнымалыларды және өрнектерді тікелей жазуға болады. Бұл — мәтінді айнымалылармен біріктірудің ең ыңғайлы тәсілі.

```python
product = "Notebook"
cost = 1250.5
quantity = 3
```

```python
# Айнымалыларды мәтін ішіне қою
print(f"Тауар: {product}, саны: {quantity}")
```

```python
# Фигуралық жақша ішінде өрнекті есептеу және форматтау
print(f"Жалпы сома: {cost * quantity:.2f} теңге")
```

## Логикалық тип және салыстыру

**Мақсаты:** логикалық мәндерді, салыстыру және логикалық операторларды түсіну, өйткені олар шартты операторлардың негізі болып табылады.

Логикалық тип `bool` тек екі мән қабылдайды: `True` (ақиқат) және `False` (жалған). Салыстыру операторлары (`==`, `!=`, `>`, `<`, `>=`, `<=`) екі мәнді салыстырып, логикалық мән қайтарады. Логикалық операторлар `and`, `or`, `not` бірнеше шартты біріктіруге мүмкіндік береді.

### Салыстыру операторлары

Назар аударыңыз: `=` мән меншіктеу операторы, ал `==` екі мәннің теңдігін тексеру операторы.

```python
x = 10
y = 7
```

```python
print(x == y)   # тең бе
print(x != y)   # тең емес пе
```

```python
print(x > y)    # үлкен бе
print(x < y)    # кіші ме
print(x >= 10)  # үлкен немесе тең бе
print(y <= 5)   # кіші немесе тең бе
```

```python
# Жолдарды да салыстыруға болады
print("python" == "python")
print("Python" == "python")   # регистр ескеріледі
```

### Логикалық операторлар

`and` екі шарт та ақиқат болса ғана `True` қайтарады. `or` шарттардың кем дегенде біреуі ақиқат болса `True` қайтарады. `not` логикалық мәнді керісінше өзгертеді.

```python
age = 15
has_id = True
```

```python
# and: екі шарт та орындалуы керек
print(age >= 18 and has_id)
```

```python
# or: кем дегенде бір шарт орындалса жеткілікті
print(age < 16 or has_id)
```

```python
# not: мәнді керісінше өзгертеді
print(not has_id)
```

```python
# Аралыққа тиесілілікті қысқаша тексеру
print(18 <= age <= 25)
```

## Шартты операторлар

**Мақсаты:** шартқа байланысты бағдарламаның әртүрлі бөлігін орындауды, яғни бағдарламада шешім қабылдауды үйрену.

Осы уақытқа дейін код жолдары ретімен түгел орындалды. Шартты операторлар `if`, `elif`, `else` белгілі бір шарт ақиқат болғанда ғана сәйкес код блогын орындауға мүмкіндік береді. Python-да код блогы шегініс (әдетте 4 бос орын) арқылы белгіленеді, ал шарттан кейін міндетті түрде қос нүкте `:` қойылады.

### if операторы

Шарт `True` болса, шегініспен жазылған блок орындалады. Шарт `False` болса, блок өткізіп жіберіледі.

```python
temperature = 15

if temperature > 25:
    print("Күн ыстық")
    print("Көп су ішіңіз")


print("Бұл жол шартқа қарамастан орындалады")
```

### if-else құрылымы

`else` блогы шарт жалған болған жағдайда орындалады. Осылайша бағдарлама екі мүмкін жолдың бірін таңдайды.

```python
number = 7

if number % 2 == 0:
    print(number, "- жұп сан")
else:
    print(number, "- тақ сан")
```

### if-elif-else құрылымы

Екіден көп нұсқа болғанда `elif` қолданылады. Шарттар жоғарыдан төмен қарай тексеріледі және бірінші ақиқат шарттың блогы ғана орындалады.

```python
score = 20
```

```python
# Баллға сәйкес баға қою
if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 50:
    grade = "C"
else:
    grade = "F"

print(f"Балл: {score}, баға: {grade}")
```

Шарттардың реті маңызды. Егер `score >= 50` шарты бірінші тұрса, 95 балл да "C" бағасын алып қалар еді, өйткені бұл шарт бірінші болып орындалады.

### Шағын мысал: дене салмағының индексі

Дене салмағының индексі салмақты бойдың квадратына бөлу арқылы есептеледі. Нәтижеге қарай шартты оператор санатты анықтайды.

```python
weight = 68     # килограмм
height = 1.75   # метр

bmi = weight / height ** 2
print("Индекс:", round(bmi, 1))
```

```python
if bmi < 18.5:
    category = "салмақ жетіспейді"
elif bmi < 25:
    category = "қалыпты салмақ"
elif bmi < 30:
    category = "артық салмақ"
else:
    category = "семіздік"

print("Санаты:", category)
```

## Циклдер

**Мақсаты:** бір әрекетті бірнеше рет қайталау үшін `for` және `while` циклдерін қолдануды, сондай-ақ циклді басқаруды үйрену.

Цикл — белгілі бір код блогын қайталап орындайтын құрылым. `for` циклі тізбектің (тізімнің, жолдың, `range()` нәтижесінің) әр элементі бойынша кезекпен өтеді. `while` циклі берілген шарт ақиқат болғанша қайталана береді. Цикл денесі де шартты операторлардағыдай шегініспен жазылады.

### Цикл не үшін қажет

1-ден 5-ке дейінгі сандарды шығару керек деп есептейік. Циклсіз әр жолды бөлек жазуға тура келеді, ал сандар мыңға жетсе, бұл тәсіл мүлдем жарамсыз болады.

```python
# Циклсіз: әр әрекетті қолмен қайталау
print(1)
print(2)
print(3)
print(4)
print(5)
```

```python
# Циклмен: дәл сол нәтиже екі жолда
for i in range(1, 6):
    print(i)
```

### range() функциясы

`range(stop)` 0-ден `stop`-қа дейінгі (`stop` кірмейді) сандар тізбегін жасайды. `range(start, stop)` басталатын санды, ал `range(start, stop, step)` қадамды да көрсетуге мүмкіндік береді.

```python
# 0-ден 4-ке дейін
for i in range(5):
    print(i, end="/")
```

```python
# 2-ден 6-ға дейін
for i in range(2, 7):
    print(i, end=" ")
```

```python
# 0-ден 20-ға дейін 5 қадаммен
for i in range(0, 26, 5):
    print(i, end=" ")
```

```python
# Кері санау: теріс қадам
for i in range(5, 0, -1):
    print(i, end=" ")
```

### Тізбектер бойынша for циклі

`for` циклі тізімнің, жолдың немесе сөздіктің элементтері бойынша тікелей өте алады. Бұл жағдайда индекстерді қолмен басқарудың қажеті жоқ.

```python
# Тізім элементтері бойынша өту
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("Жеміс:", fruit)
```

```python
# Жолдың әр символы бойынша өту
for letter in "Python":
    print(letter)
```

```python
# Сөздіктің кілттері мен мәндері бойынша өту
student = {"name": "Nurlan", "age": 19, "group": "IT-21"}

for key, value in student.items():
    print(f"{key}: {value}")
```

`enumerate()` функциясы элементпен бірге оның реттік нөмірін де береді.

```python
# Элементті және оның нөмірін бірге алу
for index, fruit in enumerate(fruits, start=1):
    print(index, fruit)
```

### Жинақтаушы айнымалы

Циклдерде жиі қолданылатын тәсіл — цикл алдында бастапқы мәні бар айнымалы құрып, әр қадамда оны жаңарту. Осылайша қосындыны, санды немесе жаңа тізімді жинақтауға болады.

```python
# 1-ден 100-ге дейінгі сандардың қосындысы
total = 0

for n in range(1, 101):
    total += n

print("Қосынды:", total)
```

```python
# Тізімдегі жұп сандарды санау
numbers = [3, 8, 12, 7, 6, 15, 4]
even_count = 0

for n in numbers:
    if n % 2 == 0:
        even_count += 1

print("Жұп сандар саны:", even_count)
```

```python
# Шартқа сай элементтерден жаңа тізім жасау
big_numbers = []

for n in numbers:
    if n > 6:
        big_numbers.append(n)

print(big_numbers)
```

### while циклі

`while` циклі қайталау саны алдын ала белгісіз болғанда қолданылады. Цикл денесінде шартқа әсер ететін айнымалыны міндетті түрде өзгерту керек, әйтпесе цикл ешқашан тоқтамайды.

```python
count = 1

while count <= 5:
    print("Қадам", count)
    count += 1   # count мәнін 1-ге арттыру
```

```python
# Салым сомасы екі есе өскенше неше жыл қажет екенін есептеу
deposit = 100000
rate = 0.12   # жылдық 12 пайыз
years = 0

while deposit < 200000:
    deposit = deposit * (1 + rate)
    years += 1

print("Қажетті жыл саны:", years)
print("Соңғы сома:", round(deposit, 2))
```

### Циклді басқару: break және continue

`break` операторы циклді толық тоқтатады. `continue` операторы ағымдағы қадамның қалған бөлігін өткізіп жіберіп, келесі қадамға көшеді.

```python
# break: бірінші теріс санды тапқанда тоқтау
values = [4, 9, 2, -3, 7, -1]

for v in values:
    if v < 0:
        print("Бірінші теріс сан:", v)
        break
```

```python
# continue: теріс сандарды өткізіп жіберу
for v in values:
    if v < 0:
        continue
    print(v, end=" ")
```

```python
# while True және break: шексіз циклден шығу
n = 1

while True:
    if n * n > 50:
        break
    n += 1

print("Квадраты 50-ден асатын ең кіші сан:", n)
```

### Кірістірілген циклдер

Бір циклдің ішінде басқа цикл жазуға болады. Сыртқы циклдің әр қадамында ішкі цикл толық орындалады.

```python
# Көбейту кестесі (1-ден 5-ке дейін)
for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end="\t")
    print()
```

### Тізімдік өрнек (list comprehension)

Тізімдік өрнек — `for` циклі арқылы жаңа тізім құрудың қысқа жазбасы. Ол жоғарыда қарастырылған жинақтаушы тізім тәсілін бір жолда жазуға мүмкіндік береді.

```python
# 1-ден 10-ға дейінгі сандардың квадраттары
squares = [n ** 2 for n in range(1, 11)]
print(squares)
```

```python
# Шарты бар тізімдік өрнек: тек жұп сандар
evens = [n for n in range(1, 21) if n % 2 == 0]
print(evens)
```

## Функциялар

**Мақсаты:** өз функцияларыңды құруды, оларға параметр беруді, нәтиже қайтаруды және функцияларды қайта қолдануды үйрену.

Функция — белгілі бір тапсырманы орындайтын, атауы бар және қайта қолдануға болатын код блогы. Біз бұған дейін `print()`, `len()`, `type()` сияқты кірістірілген функцияларды қолдандық, енді өз функцияларымызды жазамыз. Функция `def` кілт сөзімен анықталады, ал `return` операторы оның нәтижесін қайтарады.

### Функцияны анықтау және шақыру

Функцияны анықтау оны орындамайды, тек есте сақтайды. Функцияның денесі ол атауы мен жақша арқылы шақырылғанда ғана орындалады.

```python
# Функцияны анықтау
def greet():
    print("Сәлем, студент!")
```

```python
# Функцияны шақыру
greet()
greet()
```

### Параметрлер мен аргументтер

Параметр — функцияны анықтағанда жақша ішінде көрсетілетін айнымалы. Аргумент — функцияны шақырғанда параметрге берілетін нақты мән.

```python
# name — параметр
def greet_person(name):
    print(f"Сәлем, {name}!")
```

```python
# "Aigerim" және "Nurlan" — аргументтер
greet_person("Aigerim")
greet_person("Nurlan")
```

### Мән қайтару: return

`print()` нәтижені тек экранға шығарады, ал `return` нәтижені функцияны шақырған жерге қайтарады. Қайтарылған мәнді айнымалыда сақтауға және әрі қарай есептеулерде қолдануға болады.

```python
def add(a, b):
    return a + b
```

```python
# Қайтарылған мәнді айнымалыда сақтау
result = add(4, 522)
print("Нәтиже:", result)
```

```python
# Қайтарылған мәнді басқа өрнекте тікелей қолдану
print(add(10, 20) * 2)
```

`print()` пен `return` айырмашылығын көрсететін мысал: `return` жоқ функция `None` мәнін қайтарады.

```python
def add_print(a, b):
    print(a + b)

value = add_print(2, 3)
print("Қайтарылған мән:", value)
```

`return` орындалған сәтте функция жұмысын бірден аяқтайды. Сондықтан оны шартпен бірге қолданып, түрлі жағдайда түрлі нәтиже қайтаруға болады.

```python
def check_number(n):
    if n > 0:
        return "оң сан"
    elif n < 0:
        return "теріс сан"
    return "нөл"
```

```python
print(check_number(5))
print(check_number(-3))
print(check_number(0))
```

### Бірнеше мән қайтару

Функция үтір арқылы бірнеше мән қайтара алады. Бұл мәндер кортеж түрінде қайтарылады және оларды бірден бірнеше айнымалыға бөліп алуға болады.

```python
def min_max(numbers):
    return min(numbers), max(numbers)
```

```python
low, high = min_max([7, 2, 9, 4])
print("Ең кіші:", low)
print("Ең үлкен:", high)
```

### Әдепкі мәндері бар параметрлер

Параметрге әдепкі мән берілсе, функцияны шақырғанда ол аргументті көрсетпеуге болады. Әдепкі мәні бар параметрлер әрқашан соңында жазылады.

Аргументтерді параметр атауымен беруге болады. Бұл жағдайда аргументтердің реті маңызды емес және код түсініктірек болады.

### Айнымалылардың көріну аймағы

Функция ішінде құрылған айнымалы локалды айнымалы деп аталады: ол тек сол функцияның ішінде бар және сырттан көрінбейді. Функциядан тыс құрылған айнымалы глобалды болып саналады және оны функция ішінде оқуға болады.

### Функцияларды бірге қолдану

Бір функция екінші функцияны шақыра алады. Күрделі есепті шағын функцияларға бөлу кодты түсінікті әрі тексеруге ыңғайлы етеді.

```python
def average(numbers):
    return sum(numbers) / len(numbers)
```

```python
def letter_grade(score):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 50:
        return "C"
    return "F"
```

```python
# Бір функцияның нәтижесін екінші функцияға беру
scores = [88, 92, 79, 95]
avg = average(scores)
print(f"Орташа: {avg:.1f}, баға: {letter_grade(avg)}")
```

### Функция ішінде цикл мен шартты қолдану

```python
def is_prime(n):
    """n жай сан болса True, әйтпесе False қайтарады."""
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False
    return True
```

```python
print(is_prime(7))
print(is_prime(12))
```

```python
# Функцияны тізімдік өрнекпен бірге қолдану
primes = [n for n in range(1, 50) if is_prime(n)]
print(primes)
```

## Барлығын біріктіру: студенттер үлгерімін талдау

**Мақсаты:** ноутбукта үйренген барлық ұғымдарды бір практикалық есепте бірге қолдану.

Бізде студенттердің аттары мен баллдары сақталған сөздік бар. Мақсатымыз — әр студенттің орташа баллы мен бағасын анықтау, топтың жалпы көрсеткіштерін есептеу және нәтижені кесте түрінде шығару. Есеп шағын функцияларға бөлініп, қадам-қадаммен шешіледі.

### Бастапқы деректер

```python
# Кілт — студенттің аты, мән — баллдар тізімі
journal = {
    "Aigerim": [85, 92, 78, 90],
    "Nurlan": [65, 70, 58, 72],
    "Dana": [95, 98, 91, 97],
    "Arman": [45, 52, 40, 48],
    "Madina": [78, 81, 85, 80]
}
```

### Көмекші функциялар

Алдымен орташа мәнді есептейтін және бағаны анықтайтын функцияларды дайындаймыз. Олар жоғарыдағы бөлімде анықталған функцияларға ұқсас, бірақ мұнда есеп толық болуы үшін қайта жазылған.

```python
def calc_average(scores):
    """Баллдар тізімінің орташа мәнін қайтарады."""
    return sum(scores) / len(scores)
```

```python
def get_grade(score):
    """Баллға сәйкес әріптік бағаны қайтарады."""
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 50:
        return "C"
    return "F"
```

### Әр студент бойынша нәтиже

Цикл арқылы сөздіктегі әр студент бойынша өтіп, оның орташа баллы мен бағасын есептейміз. Нәтижелерді кейін қолдану үшін жаңа сөздікке жинаймыз.

```python
results = {}

for name, scores in journal.items():
    avg = calc_average(scores)
    results[name] = avg
```

```python
# Нәтижені кесте түрінде шығару
print(f"{'Студент':<10}{'Орташа':>8}{'Баға':>6}")
print("-" * 24)

for name, avg in results.items():
    print(f"{name:<10}{avg:>8.1f}{get_grade(avg):>6}")
```

### Топтың жалпы көрсеткіштері

```python
# Топтың орташа баллы
group_average = calc_average(list(results.values()))
print(f"Топтың орташа баллы: {group_average:.1f}")
```

```python
# Ең жоғары баллы бар студент
best_student = max(results, key=results.get)
print("Үздік студент:", best_student)
```

```python
# Емтиханнан өтпеген студенттер (баға F)
failed = [name for name, avg in results.items() if get_grade(avg) == "F"]
print("Қайта тапсыруы тиіс:", failed)
```

```python
# Әр баға бойынша студенттер санын есептеу
grade_counts = {}

for avg in results.values():
    grade = get_grade(avg)
    grade_counts[grade] = grade_counts.get(grade, 0) + 1

print(grade_counts)
```

## Өзіндік жұмысқа арналған тапсырмалар

**Мақсаты:** өткен материалды өз бетінше қолдану арқылы бекіту.

Әр тапсырманы астындағы бос ұяшықта шешіңіз. Тапсырмалар ноутбуктағы бөлімдердің ретімен берілген және біртіндеп күрделене түседі.

**Тапсырма 1.** `city` және `population` айнымалыларын құрып, оларды f-жол арқылы бір сөйлем етіп шығарыңыз.

```python
# Шешіміңізді осында жазыңыз
city = "Almaty"
population = 2388000
print(f"The population of the city of {city} today is {population} people.")
```

**Тапсырма 2.** Цельсий бойынша берілген температураны Фаренгейтке айналдырыңыз. Формула: F = C * 9 / 5 + 32.

```python
# Шешіміңізді осында жазыңыз
C = int(input())
F = C * 9/5 + 32
print(F)
```

**Тапсырма 3.** Берілген жылдың кібісе жыл екенін анықтаңыз. Жыл 4-ке бөлінсе және 100-ге бөлінбесе, немесе 400-ге бөлінсе, ол кібісе жыл болады.

```python
# Шешіміңізді осында жазыңыз
year = int(input())

if year % 4 == 0 and year % 100 != 0 or year % 400 == 0:
  print(year, "кібісе жыл")
else:
  print(year, "кібісе жыл емес")
```

**Тапсырма 4.** `for` циклі арқылы 1-ден 50-ге дейінгі 3-ке де, 5-ке де бөлінетін сандарды шығарыңыз.

```python
# Шешіміңізді осында жазыңыз
for i in range (1, 50):
  if i % 3 == 0 and i % 5 == 0:
    print(i)
```

**Тапсырма 5.** Жолдағы дауысты әріптердің (a, e, i, o, u) санын қайтаратын `count_vowels(text)` функциясын жазыңыз.

```python
# Шешіміңізді осында жазыңыз
text = input()
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for letter in text:
        if letter in vowels:
            count += 1

    return count
print(count_vowels(text))
```

**Тапсырма 6.** Сандар тізімін қабылдап, оның тек оң элементтерінен тұратын жаңа тізімді қайтаратын `filter_positive(numbers)` функциясын жазыңыз. Функцияны цикл арқылы, содан кейін тізімдік өрнек арқылы жазып көріңіз.

```python
# Шешіміңізді осында жазыңыз
def filter_positive(numbers):
    my_list = [i for i in numbers if i >= 0]
    return my_list

print(filter_positive([-2, -1, 0, 1, 2]))
```

## Қорытынды

Бұл ноутбукта Python бағдарламалау тілінің негізгі ұғымдары қарапайымнан күрделіге қарай біртіндеп қарастырылды. Айнымалылар мәндерді сақтауға, деректер типтері олармен қандай амал жасауға болатынын анықтауға, ал тізімдер мен сөздіктер көп деректі бір құрылымда ұйымдастыруға мүмкіндік береді.

Шартты операторлар бағдарламаға шешім қабылдауды, циклдер әрекеттерді қайталауды, ал функциялар кодты қайта қолдануға болатын логикалық бөліктерге бөлуді қамтамасыз етеді. Соңғы практикалық мысал бұл ұғымдардың бір-бірімен тығыз байланысты екенін және бірге қолданылғанда нақты есептерді шешуге болатынын көрсетті.

Материалды бекіту үшін мысалдардағы мәндерді өзгертіп, ұяшықтарды қайта орындаңыз және өзіндік жұмыс тапсырмаларын толық орындаңыз.
